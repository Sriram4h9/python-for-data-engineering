* (*) astrick will take those values and stored in tuple

```python
def my_func(*p_x):

    print(p_x)

my_func(40,50,60,70,80,90,100)
```

```python
def my_func(*p_x):

    print(p_x)
    print(type(p_x))

my_func(40,50,60,70,80,90,100)
```

* (**) It will convert into dictionary

```python
def my_func(**p_x):

    print(p_x)
    print(type(p_x))
    print(p_x.keys())
    print(p_x.values())
    print(p_x.items())

my_func(x=10,y=20,z=30)
```

* Map
* Filter
* Reduce

```python
my_list = [1,2,3,4,5]

def square(p_x):

    ret_list = [i*i for i in p_x]

    print(ret_list)
square(my_list)
```

```python
my_list = [1,2,3,4,5]

def square(p_x):

    return p_x*p_x
result = list(map(square,my_list))

print(result)
```

```python
my_list = [1,2,3,4,5]

def square(p_x):

    if (p_x % 2 == 0):
        return p_x*p_x

result = list(map(square,my_list))

print(result)
```

```python
my_list = [1,2,3,4,5]

def square(p_x):

    if (p_x % 2 == 0):
        return p_x*p_x

result = list(filter(square,my_list))

print(result)
```

from functools import reduce -- to import the reduce function 

```python
from functools import reduce
my_list = [1,2,3,4,5]

def square(p_x,p_y):

    return p_x+p_y

result = reduce(square,my_list)

print(result)
```

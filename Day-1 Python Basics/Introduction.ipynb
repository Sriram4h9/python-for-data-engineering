### List Comprehension 

```python
LIST = [1,2,3,4,5,6]

new_list = []

for i in LIST:
    new_list.append(i*i)
print(new_list)
```

```python
LIST = [1,2,3,4,5,6]

new_list = [i*i for i in LIST]
print(new_list)
```

```python
LIST = [1,2,3,4,5,6]

new_list = [i*i for i in LIST if (i%2 == 0)]
print(new_list)
```

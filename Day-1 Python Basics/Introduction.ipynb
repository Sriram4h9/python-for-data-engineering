# Comments in Python 

**Single Line Comment**  (#)\
**Multiline Line Comment** (''' ''')

```python
# 10 value is stored in the variable called var - This is a Single line comment

Var1 = 10

# my name - This is a Single line comment

var2 = "Sriram"
```

```python

''' **This is a Multiline comment**
I am preparing Data Engineer along with AI.
I am Starting to Learn Python from Scrach.
Happy to Learn.
'''
```

```python
# Multiline Code 

multi_line = 10+20+30+40
print(multi_line)

 # ---------- or -----------

multi_lines = 10+20\
              +30+40
print(multi_lines)

 # ---------- or -----------

multi_liness = (10+20
                +30+40)
print(multi_liness)
```
